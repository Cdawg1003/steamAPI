```python
text_corpus = df['review'].to_list()
 
```

```python
import re

text_corpus = [re.sub(r'\W', ' ', i) for i in text_corpus]
```

```python
from gensim import corpora
```

```python
import pprint
from gensim import corpora 

#the document variable is a dummy string because I am unsure if a document needs to be included or if we can only include a corpus (list of documents)
document = 'This game is HOT garbage. All of the player base is a bunch of hackers.'

import nltk

#nltk.download("stopwords")

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

#nltk.download('punkt')

stoplist = stopwords.words('english')
stoplist.append('game')

texts = [[word for word in document.lower().split() if word not in stoplist]
         for document in text_corpus]

from collections import defaultdict
frequency = defaultdict(int)
for text in texts: 
    for token in text: 
        frequency[token] += 1

processed_corpus = [[token for token in text if frequency[token] > 1] for text in texts]
 
```

```python
from gensim import corpora

dictionary = corpora.Dictionary(processed_corpus)
print(dictionary)
```

```python
from nltk.stem.wordnet import WordNetLemmatizer 

lemmatizer = WordNetLemmatizer()

docs = [[lemmatizer.lemmatize(token) for token in processed_corpus] for processed_corpus in processed_corpus]
```

```python
from gensim.models import Phrases

bigram = Phrases(processed_corpus, min_count=20)
for idx in range(len(processed_corpus)):
    for token in bigram[processed_corpus[idx]]:
        if '_' in token: 
            docs[idx].append(token)
```

```python
print(dictionary.token2id)
```

```python
bow_corpus = [dictionary.doc2bow(text) for text in processed_corpus]
print(bow_corpus)
```

```python
from gensim.models import LdaModel

num_topics = 5
chunksize = 5000
passes = 50 
iterations = 1000
eval_every = None

temp = dictionary[0]
id2word = dictionary.id2token

lda_model = LdaModel(
    corpus = bow_corpus, 
    id2word=dictionary, 
    chunksize=chunksize, 
    alpha = 'auto', 
    iterations=iterations, 
    num_topics=num_topics, 
    passes=passes, 
    eval_every=eval_every)
```

```python
pprint.pprint(lda_model.print_topics())
```
